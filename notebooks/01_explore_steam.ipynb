```python
import pandas as pd

df = pd.read_parquet("../data/raw/steam_games.parquet")

print(df.shape)       # (rows, columns)
print(df.dtypes)      # each column and its data type
```

```python
print(df.dtypes.to_string())
```

```python
df.head(3).T
```

```python
# share of games where each number column is 0
num_cols = ["price", "positive", "negative", "metacritic_score",
            "user_score", "peak_ccu", "average_playtime_forever"]
print("Share of zeros:")
print((df[num_cols] == 0).mean().round(3).to_string())

# share of games where each list column is empty
print("\nShare of empty lists:")
for col in ["genres", "tags", "developers", "categories"]:
    empty = df[col].apply(lambda x: x is None or len(x) == 0).mean()
    print(f"{col}: {empty:.3f}")

# how many entries look like playtests
print("\nPlaytest entries:", df["name"].str.contains("playtest", case=False, na=False).sum())
```

```python
df["total_reviews"] = df["positive"] + df["negative"]

for t in [1, 10, 50, 100, 500]:
    print(f">= {t} reviews: {(df['total_reviews'] >= t).sum():,} games")
```
